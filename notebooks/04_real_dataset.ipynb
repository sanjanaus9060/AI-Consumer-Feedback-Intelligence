```python
import gzip
import pandas as pd
```

```python
file_path = "../data/raw/finefoods.txt.gz"

with gzip.open(file_path, "rt", encoding="latin-1") as f:
    lines = f.readlines()

print("Total lines:", len(lines))
```

```python
reviews = []
current_review = {}

for line in lines:
    line = line.strip()

    if line.startswith("review/score:"):
        current_review["rating"] = float(line.split(":", 1)[1].strip())

    elif line.startswith("review/text:"):
        current_review["review"] = line.split(":", 1)[1].strip()

        if "rating" in current_review:
            reviews.append(current_review.copy())

        current_review = {}

print("Total reviews extracted:", len(reviews))
```

```python
df_real = pd.DataFrame(reviews)

print(df_real.head())
print("\nDataset shape:", df_real.shape)
```

```python
def create_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"


df_real["sentiment"] = df_real["rating"].apply(create_sentiment)

print(df_real["sentiment"].value_counts())
```

```python
positive_df = df_real[df_real["sentiment"] == "Positive"].sample(
    n=40000, random_state=42
)

negative_df = df_real[df_real["sentiment"] == "Negative"].sample(
    n=40000, random_state=42
)

neutral_df = df_real[df_real["sentiment"] == "Neutral"].sample(
    n=40000, random_state=42
)

df_balanced = pd.concat(
    [positive_df, negative_df, neutral_df],
    ignore_index=True
)

# Shuffle the dataset
df_balanced = df_balanced.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print("Balanced dataset shape:", df_balanced.shape)
print("\nSentiment distribution:")
print(df_balanced["sentiment"].value_counts())
```

```python
output_path = "../data/processed/real_balanced_reviews.csv"

df_balanced.to_csv(output_path, index=False)

print("Balanced dataset saved successfully!")
print("File:", output_path)
```

```python
import gzip
import pandas as pd

file_path = "../data/raw/finefoods.txt.gz"

reviews = []
current_review = {}

with gzip.open(file_path, "rt", encoding="latin-1") as f:
    for line in f:

        line = line.strip()

        if line.startswith("review/score:"):
            current_review["rating"] = float(
                line.split(":", 1)[1].strip()
            )

        elif line.startswith("review/text:"):
            current_review["review"] = line.split(":", 1)[1].strip()

            if "rating" in current_review:
                reviews.append(current_review.copy())

            current_review = {}

print("Total reviews extracted:", len(reviews))
```

```python
df_real = pd.DataFrame(reviews)

print("Dataset shape:", df_real.shape)
print("\nColumns:")
print(df_real.columns.tolist())

print("\nFirst 5 rows:")
print(df_real.head())
```

```python
def create_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"


df_real["sentiment"] = df_real["rating"].apply(create_sentiment)

print("Sentiment distribution:")
print(df_real["sentiment"].value_counts())
```

```python
df_real.to_csv(
    "../data/processed/real_reviews.csv",
    index=False
)

print("Large dataset saved successfully!")
```

```python
df_balanced = pd.concat([
    df_real[df_real["sentiment"] == "Positive"].sample(
        n=40000, random_state=42
    ),
    df_real[df_real["sentiment"] == "Negative"].sample(
        n=40000, random_state=42
    ),
    df_real[df_real["sentiment"] == "Neutral"].sample(
        n=40000, random_state=42
    )
], ignore_index=True)

print("Balanced dataset shape:", df_balanced.shape)

print("\nSentiment distribution:")
print(df_balanced["sentiment"].value_counts())
```

```python
df_balanced.to_csv(
    "../data/processed/real_balanced_reviews.csv",
    index=False
)

print("Balanced dataset saved successfully!")
```
